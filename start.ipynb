# 내 역할부터 실행하기

현재 코드는 Physical AI 분야의 **가상 기업 연결 예제**입니다. 실제 LLM·RAG는 아직 구현하지 않았으며, 예제 실행에 API 키나 모델 다운로드는 필요하지 않습니다.

먼저 [README의 Usage](README.md#usage)를 따라 프로젝트 안에 `.venv`를 만들고 패키지를 설치하세요. 이 노트북의 커널은 **그 `.venv`의 Python 3.11**로 선택합니다. 기존 수업 환경은 그대로 둡니다. VS Code에서 커널 사용에 필요한 경우, 활성화한 `.venv`에 `python -m pip install ipykernel`을 실행하세요(노트북 사용자만 필요).

노트북과 `main.py`가 있는 저장소 폴더를 작업 폴더로 열고 아래 셀을 순서대로 실행하세요.


```python
from pathlib import Path
import subprocess
import sys

project_dir = Path.cwd()
if not (project_dir / "main.py").is_file():
    raise RuntimeError("main.py가 있는 저장소 폴더에서 이 노트북을 열어주세요.")
print("사용 중인 Python:", sys.executable)
print("작업 폴더:", project_dir)
```

## 1. 내 역할만 시험하기

`agents/technology.py`의 `run(state)`를 구현한 뒤 아래 셀을 다시 실행하면 됩니다. 앞 담당자의 결과는 `data/test_state.json`의 예제로 받으므로 기다릴 필요가 없습니다.

담당에 따라 `technology`를 `company`, `market_competition`, `investment`, `report`로 바꾸세요. 나머지 네 역할은 예제로 유지해도 됩니다.

다른 입력을 쓰려면 목록 끝에 `"--state", "data/my_test_state.json"`을 추가합니다. 결과 폴더를 나누려면 `"--output", "outputs/my-test"`를 추가합니다.


```python
subprocess.run([sys.executable, "main.py", "--agent", "technology"], check=True)
```

## 2. 전체 연결 확인하기

전체 실행은 후보마다 기업 정보 → 기술 → 시장·경쟁사 → 투자 판단 순서로 진행합니다. 보류하면 다음 후보로 넘어가고, 투자할 후보를 찾거나 모든 후보가 보류되면 보고서를 만듭니다.

후보는 `data/companies.json`에서 팀이 직접 정해도 됩니다. 지금은 두 가상 기업이 모두 보류되는 예제입니다. 실제 API를 연결한 함수를 실행하면 해당 호출 비용이 발생할 수 있습니다.


```python
subprocess.run([sys.executable, "main.py"], check=True)
```
