# AI 스타트업 투자평가 — 흐름 확인하기

회사탐색 → 기술분석 → 시장분석 → 경쟁분석 → 투자판단 → 보고서작성

현재는 **가상 회사 두 개로 연결을 확인하는 예제**입니다. 실제 회사는 아직 선정하지 않았고, 실제 AI·RAG 분석도 구현되지 않았습니다. 만들어지는 파일은 제출용 투자 보고서가 아닙니다.
아래 셀을 위에서부터 실행하세요. API 키나 외부 네트워크는 사용하지 않습니다.

```python
from app import build_graph, load_demo_candidates

# data/example.json에서 가상 회사 목록을 읽습니다.
candidates = load_demo_candidates()
candidates
```

```python
# 분석을 시작할 때 다음 단계에 전달할 공용 데이터입니다.
initial_state = {
    "candidates": candidates,
    "candidate_index": 0,
    "evaluations": [],
    "mode": "demo",
}
```

```python
graph = build_graph()
result = graph.invoke(initial_state, config={"recursion_limit": 50})
print("가상 자료로 흐름 확인 완료. 실제 AI 분석 결과가 아닙니다.")
```

```python
# outputs/demo에 저장된 예제 파일의 경로입니다. 제출용이 아닙니다.
result["report_paths"]
```
